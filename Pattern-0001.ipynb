```python
n = 5
for i in range(n):
    print("* " * (n))
```

```python
for i in range(5):
    print("* " * (i + 1))
    
```

```python
for i in range(1, 6):
    for j in range(1, i+1):
        print(j, end = " ")
    print()
```

```python
n = 5

for i in range(1, n + 1):
    for j in range(i):
        print(i, end = " ")
    print()
    
```

```python
rows = 5

for i in range(rows):
    print("* " * (rows - i))
    
```

```python
n = 5

for i in range(n, 0, -1):
    for j in range(i):
        print(j + 1, end = " ")
    print()
```
