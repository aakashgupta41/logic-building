```python
# Armstrong number => 
# 370 => 3^3 + 7^3 + 0^3 => 370

n = 370
original = n
total = 0
while n > 0:
    remainder = n % 10 
    cube = remainder ** 3
    total = total + cube
    n = n // 10

if original == total:
    print(f"Yes, {original} is an Armstrong Number.")
else:
    print(f"No, {original} is not an Armstrong number.")
    
```
