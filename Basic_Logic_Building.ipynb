```python
print("1. Anagram\n")

string1 = "coal"
string2 = "cola"
print(f"Comparing strings: '{string1}' and '{string2}'")

temp1 = "".join(sorted(string1))
temp2 = "".join(sorted(string2))
print(f"Sorted string 1: '{temp1}'")
print(f"Sorted string 2: '{temp2}'")

if temp1 == temp2:
    print("Result: Yes, Anagram (Both strings contain the exact same characters)")
else:
    print("Result: No")
    
```

```python
print("2. Swap 2 Numbers\n")

print("(i) Using third variable:\n")

x = 10
y = 20
print(f"Before x: {x} and y: {y}")
temp = y
y = x
x = temp
print(f"After x: {x} and y: {y}\n")

print("(ii) Without third variable:\n")
x = 10
y = 20
print(f"Before x: {x} and y: {y}")
x, y = y, x
print(f"After x: {x} and y: {y}")

print("\n(iii) Arithmetic approach:")

x = 10 
y = 20
print(f"\nBefore x: {x} and y: {y}")
x = x - y
y = y - x
x = x + y
print(f"After x: {x} and y: {y}")
```

```python
print("3. Reverse a Number\n")

n = 1452
temp = 0
print(f"Before reverse: {n}")
while n > 0:
    temp = temp * 10
    temp = temp + n % 10
    n = n // 10

print(f"After reverse: {temp}")
```

```python
print("4. Sum of Digits\n")
print("Process: 1 + 5 + 4 + 7 = 17\n")

n = 1547
original_n = n
temp = 0
print(f"Number: {n}")

while n > 0:
    remain = n % 10
    temp += remain
    n = n // 10

print(f"Sum of digits: {temp}")
```

```python
print("5. Even or Odd (Range)\n")
print(f"Process: {n}^2 = {square} => 8 + 1 = 9\n")

n = 10
even = []
odd = []
print(f"Range up to: {n}")

if n >= 0:
    for i in range(0, n + 1):
        if i % 2 == 0:
            even.append(i)
        else:
            odd.append(i)
else:
    for j in range(-1, n - 1, -1):
        if j % 2 == 0:
            even.append(j)
        else:
            odd.append(j)

print(f"Even numbers: {even}")
print(f"Odd numbers: {odd}")
```

```python
print("6. Neon Number\n")
print("Process: Reverse 121 => 1, 12, 121\n")

n = 9
square = n * n
total_sum = 0
print(f"Number: {n}, Square: {square}")

temp_sq = square
while temp_sq > 0:
    remain = temp_sq % 10
    total_sum += remain
    temp_sq //= 10

if total_sum == n:
    print(f"Result: Yes, Neon Number (Sum of digits of square {total_sum} == {n})")
else:
    print("Result: No")
    
```

```python
print("7. Palindrome Number\n")
print("Process: 1^3 + 5^3 + 3^3 = 1 + 125 + 27 = 153\n")

n = 12321
original = n
temp = 0
print(f"Number: {n}")

while n > 0:
    remain = n % 10
    temp = temp * 10 + remain
    n = n // 10

if temp == original:
    print("Result: Yes, Palindrome")
else:
    print("Result: No")
    
```

```python
print("8. Armstrong Number\n")
print("Process: 1! + 4! + 5! = 1 + 24 + 120 = 145\n")

n = 370
original = n
temp = 0
print(f"Number: {n}")

while n > 0:
    remain = n % 10
    cubes = remain ** 3
    temp += cubes
    n //= 10

if original == temp:
    print("Result: Yes, Armstrong Number")
else:
    print("Result: No")
    
```

```python
print("9. Factorial Number\n")

n = 5
temp = 1
print(f"Number: {n}")

if n == 0 or n == 1:
    print("Factorial: 1")
elif n < 0:
    print("Invalid input")
else:
    for i in range(1, n + 1):
        temp = temp * i
    print(f"Factorial: {temp}")
    
```

```python
print("10. Prime Number\n")
print("Process: 7 has only two factors: 1 and 7\n")

n = 13
flag = False
print(f"Number: {n}")

for i in range(2, n):
    if n % i == 0:
        flag = False
        break
else:
    flag = True

if flag and n > 1:
    print("Result: Prime")
else:
    print("Result: Not Prime")
    
```

```python
print("11. Fibonacci Series\n")
print("Process: 0 | 0 + 1 = 1 | 1 + 1 = 2 | 1 + 2 = 3 | 2 + 3 = 5 | ...\n")

n = 10
n1 = 0
n2 = 1
print(f"Fibonacci sequence up to {n} terms:")

for i in range(n):
    print(n1, end=" ")
    nth = n1 + n2
    n1 = n2
    n2 = nth
print()
```

```python
print("12. Perfect Number\n")
print("Process: 1 + 2 + 3 = 6 => 6 is a Perfect Number because the sum of its proper divisors equals the number itself.\n")
print("6 % 1 == 0 | 6 % 2 == 0 | 6 % 3 == 0 => 1 + 2 + 3 = 6\n")

n = 6
total = 0
print(f"Number: {n}")

for i in range(1, n):
    if n % i == 0:
        total += i

if total == n:
    print("Result: Yes, Perfect Number")
else:
    print("Result: No")
    
```

```python
print("13. Strong Number\n")
print("Process: 1! + 4! + 5! = 1 + 24 + 120 = 145\n")

n = 145
original = n
total = 0
print(f"Number: {n}")

while n > 0:
    remain = n % 10
    fact = 1
    for i in range(1, remain + 1):
        fact = fact * i
    total += fact
    n //= 10

if original == total:
    print("Result: Yes, Strong Number")
else:
    print("Result: No")
    
```

```python
print("14. Fascinating Number\n")
print("Process: 273 × 1 = 273, 273 × 2 = 546, 273 × 3 = 819 => 273546819\n")
print("273 is a Fascinating Number because 273546819 contains all digits from 1 to 9 exactly once.\n")

n = 273
print(f"Number: {n}")

# Concatenate n * 1, n * 2, n * 3
fascinating_str = str(n) + str(n * 2) + str(n * 3)
print(f"Concatenated string: {fascinating_str}")

# Check if all digits from 1 to 9 are present exactly once
is_fascinating = True
for digit in "123456789":
    if fascinating_str.count(digit) != 1:
        is_fascinating = False
        break

if is_fascinating and len(fascinating_str) == 9:
    print("Result: Yes, Fascinating Number")
else:
    print("Result: No")
    
```
